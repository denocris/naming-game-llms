```python
import matplotlib.pyplot as plt
import pandas as pd

import matplotlib.cm as cm
import numpy as np
```

```python
def success_plot(indicator):
    T = len(indicator)
    CS = [0] * (T+1)
    for i in range(1, T):
        CS[i + 1] = CS[i] + indicator[i]
    # Create cs list
    cs = [(i, CS[i]) for i in range(T)]

    # Define δ
    δ = 300

    result = [(i, (cs[i + δ][1] - cs[i][1]) / δ) for i in range(T - δ)]
    return result


def get_unique_words(settings= "na5e2-ns1e4", model_name='llama-3.2-1b-preview', temp=0.0):
    file_name = f"./results/num_of_diff_words_{settings}_{model_name}_log_T{temp}.json"
    return pd.read_json(file_name)

def get_all_words(settings= "na5e2-ns1e4", model_name='llama-3.2-1b-preview', temp=0.0):
    file_name = f"./results/num_of_words_{settings}_{model_name}_log_T{temp}.json"
    return pd.read_json(file_name)

def get_success_rate(settings= "na5e2-ns1e4",model_name='llama-3.2-1b-preview', temp=0.0):
    file_name = f"./results/success_{settings}_{model_name}_log_T{temp}.csv"
    return pd.read_csv(file_name)
```

```python
def success_plot(indicator):
    T = len(indicator)
    CS = [0] * (T+1)
    for i in range(1, T):

        CS[i + 1] = CS[i] + indicator[i]
    # Create cs list
    cs = [(i, CS[i]) for i in range(T)]

    # Define δ
    δ = 300

    result = [(i, (cs[i + δ][1] - cs[i][1]) / δ) for i in range(T - δ)]
    return result

def load_results(folder="results",type='num_of_diff_words', model_name='phi4-mini-3.8b', seed = 0, experiment = 'na5e2-ns1e4', temp=0.0):
    if type=='success':
        file_name = f"./{folder}/{type}_seed-{seed}_{experiment}_{model_name}_log_T-{temp}.csv"
        results = pd.read_csv(file_name)
    elif type=='num_of_words' or type=='num_of_diff_words':
        file_name = f"./{folder}/{type}_seed-{seed}_{experiment}_{model_name}_log_T-{temp}.json"
        results = pd.read_json(file_name)
    else:
        raise ValueError("Invalid type. Must be 'success', 'num_of_words', or 'num_of_diff_words'.")
    return results
```

```python
def get_model_answers(model_path='gemma2-2b', seed=0, steps=1000, temp=0.0):
    file_name = f"./answers/list_of_answers_{model_path}_seed-{seed}_steps-{steps}_temp-{temp}.csv"
    df = pd.read_csv(file_name)
    print("Nr. of yes:", (df['0'] == 'yes').sum())
    print("Nr. of no:", (df['0'] == 'no').sum())
    print("Nr. of others:", steps - (df['0'] == 'yes').sum() - (df['0'] == 'no').sum())
    return df


for model_path in ['phi4-mini-3.8b', 'mistral-7b', 'llama3.1-8b', 'phi3-14b', 'gemma2-2b', "gemma3n-e2b", "gemma3-270m", "qwen3-8b", "qwen3-1.7b", "llama3.2-3b"]:
    print(model_path)
    print("Low T")
    get_model_answers(model_path=model_path, seed=0, steps=300, temp=0.05)
    print("High T")
    get_model_answers(model_path=model_path, seed=0, steps=300, temp=1.95)
    print('-------------------------------')
```
