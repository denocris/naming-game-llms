```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
def success_plot(indicator):
    T = len(indicator)
    CS = [0] * (T+1)
    for i in range(1, T):
        CS[i + 1] = CS[i] + indicator[i]
    # Create cs list
    cs = [(i, CS[i]) for i in range(T)]

    # Define δ
    δ = 300

    result = [(i, (cs[i + δ][1] - cs[i][1]) / δ) for i in range(T - δ)]
    return result



def get_unique_words(settings= "na5e2-ns1e4", model_name='llama-3.2-1b', temp=0.0):
    file_name = f"./num_of_diff_words_{settings}_{model_name}_log_T{temp}.json"
    return pd.read_json(file_name)

def get_all_words(settings= "na5e2-ns1e4", model_name='llama-3.2-1b', temp=0.0):
    file_name = f"./num_of_words_{settings}_{model_name}_log_T{temp}.json"
    return pd.read_json(file_name)

def get_success_rate(settings= "na5e2-ns1e4",model_name='llama-3.2-1b', temp=0.0):
    file_name = f"./success_{settings}_{model_name}_log_T{temp}.csv"
    return pd.read_csv(file_name)
```
